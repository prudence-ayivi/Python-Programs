# Choropleth Map of African country satellite

```python
import folium

africansats_map=folium.Map(location=[2.0, 23.5], zoom_start=3)

africansats_map
```

```python
#GeoJSON File
africa_map='./africa-countries.json'

folium.Choropleth(
    geo_data=africa_map,
    fill_opacity=0.3,
    line_weight=2,
    ).add_to(africansats_map)

#map of satellite per african country

# africansats_map.choropleth(
#     geo_path=africa_map,
#     legend='African Satellites',
#     fill_color='#D7DFF2'
# )

africansats_map
```
